```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import warnings

warnings.filterwarnings("ignore")

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.cluster import KMeans
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    silhouette_score
)
```

```python
file_path = "online_retail_II.xlsx"

xls = pd.ExcelFile(file_path)

print("Available sheets:")
print(xls.sheet_names)

df1 = pd.read_excel(file_path, sheet_name=xls.sheet_names[0])
df2 = pd.read_excel(file_path, sheet_name=xls.sheet_names[1])

df = pd.concat([df1, df2], ignore_index=True)

print("Dataset shape:", df.shape)
df.head()
```

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

```python
df.describe(include="all").T
```

```python
df = df.copy()

# Convert date
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

# Remove rows with missing important information
df = df.dropna(
    subset=[
        "Invoice",
        "StockCode",
        "Description",
        "Quantity",
        "InvoiceDate",
        "Price"
    ]
)

# Remove invalid prices
df = df[df["Price"] > 0]

# Remove zero quantities
df = df[df["Quantity"] != 0]

# Identify cancellation/return invoices
df["IsReturn"] = (
    df["Invoice"]
    .astype(str)
    .str.upper()
    .str.startswith("C")
    .astype(int)
)

# Calculate revenue
df["Revenue"] = df["Quantity"] * df["Price"]

print("Cleaned shape:", df.shape)

df.head()
```

```python
df["Year"] = df["InvoiceDate"].dt.year
df["Month"] = df["InvoiceDate"].dt.month
df["Day"] = df["InvoiceDate"].dt.day
df["Hour"] = df["InvoiceDate"].dt.hour
df["DayOfWeek"] = df["InvoiceDate"].dt.dayofweek

df["MonthName"] = df["InvoiceDate"].dt.strftime("%b")

print(df.head())
```

```python
df["Date"] = df["InvoiceDate"].dt.date
daily = (
    df.groupby("Date")
    .agg(
        DailyRevenue=("Revenue", "sum"),
        DailyQuantity=("Quantity", "sum"),
        AveragePrice=("Price", "mean"),
        Transactions=("Invoice", "nunique"),
        Returns=("IsReturn", "sum")
    )
    .reset_index()
)

daily["Date"] = pd.to_datetime(daily["Date"])

daily = daily.sort_values("Date").reset_index(drop=True)

print("Daily dataset shape:", daily.shape)

daily.head()
```

```python
# Previous day revenue
daily["PreviousRevenue"] = daily["DailyRevenue"].shift(1)

# Previous day quantity
daily["PreviousQuantity"] = daily["DailyQuantity"].shift(1)

# Rolling 7-day revenue
daily["Rolling7Revenue"] = (
    daily["DailyRevenue"]
    .shift(1)
    .rolling(7)
    .mean()
)

# Rolling 7-day quantity
daily["Rolling7Quantity"] = (
    daily["DailyQuantity"]
    .shift(1)
    .rolling(7)
    .mean()
)

# Date features
daily["Year"] = daily["Date"].dt.year
daily["Month"] = daily["Date"].dt.month
daily["Day"] = daily["Date"].dt.day
daily["DayOfWeek"] = daily["Date"].dt.dayofweek

# Future revenue = next day's revenue
daily["FutureRevenue"] = daily["DailyRevenue"].shift(-1)

# Remove rows created by shift/rolling operations
daily = daily.dropna().reset_index(drop=True)

print("Daily dataset:", daily.shape)

daily.head()
```

```python
regression_features = [
    "PreviousRevenue",
    "PreviousQuantity",
    "Rolling7Revenue",
    "Rolling7Quantity",
    "AveragePrice",
    "Transactions",
    "Returns",
    "Month",
    "DayOfWeek"
]

X_reg = daily[regression_features]
y_reg = daily["FutureRevenue"]

print("Number of regression features:", len(regression_features))
print(regression_features)
```

```python
split_index = int(len(daily) * 0.8)

X_train_reg = X_reg.iloc[:split_index]
X_test_reg = X_reg.iloc[split_index:]

y_train_reg = y_reg.iloc[:split_index]
y_test_reg = y_reg.iloc[split_index:]

print("Training rows:", len(X_train_reg))
print("Testing rows:", len(X_test_reg))
```

```python
linear_model = LinearRegression()

linear_model.fit(
    X_train_reg,
    y_train_reg
)

linear_pred = linear_model.predict(X_test_reg)

linear_mae = mean_absolute_error(
    y_test_reg,
    linear_pred
)

linear_rmse = np.sqrt(
    mean_squared_error(
        y_test_reg,
        linear_pred
    )
)

linear_r2 = r2_score(
    y_test_reg,
    linear_pred
)

print("Linear Regression")
print("-----------------")
print("MAE:", linear_mae)
print("RMSE:", linear_rmse)
print("R2:", linear_r2)
```

```python
rf_regressor = RandomForestRegressor(
    n_estimators=150,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)

rf_regressor.fit(
    X_train_reg,
    y_train_reg
)

rf_pred = rf_regressor.predict(X_test_reg)

rf_mae = mean_absolute_error(
    y_test_reg,
    rf_pred
)

rf_rmse = np.sqrt(
    mean_squared_error(
        y_test_reg,
        rf_pred
    )
)

rf_r2 = r2_score(
    y_test_reg,
    rf_pred
)

print("Random Forest Regression")
print("------------------------")
print("MAE:", rf_mae)
print("RMSE:", rf_rmse)
print("R2:", rf_r2)
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    y_test_reg.values[:100],
    label="Actual"
)

plt.plot(
    rf_pred[:100],
    label="Predicted"
)

plt.title("Actual vs Predicted Future Revenue")
plt.xlabel("Test Observations")
plt.ylabel("Future Revenue")

plt.legend()

plt.show()
```

```python
daily["FutureQuantity"] = daily["DailyQuantity"].shift(-1)

median_quantity = daily["FutureQuantity"].median()

daily["HighDemand"] = (
    daily["FutureQuantity"] > median_quantity
).astype(int)

daily = daily.dropna().reset_index(drop=True)

print(
    daily["HighDemand"].value_counts()
)
```

```python
classification_features = [
    "PreviousRevenue",
    "PreviousQuantity",
    "Rolling7Revenue",
    "Rolling7Quantity",
    "AveragePrice",
    "Transactions",
    "Returns",
    "Month",
    "DayOfWeek"
]

X_cls = daily[classification_features]
y_cls = daily["HighDemand"]

split_index_cls = int(len(daily) * 0.8)

X_train_cls = X_cls.iloc[:split_index_cls]
X_test_cls = X_cls.iloc[split_index_cls:]

y_train_cls = y_cls.iloc[:split_index_cls]
y_test_cls = y_cls.iloc[split_index_cls:]

print("Classification features:", len(classification_features))
```

```python
logistic_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced"
)

logistic_model.fit(
    X_train_cls,
    y_train_cls
)

logistic_pred = logistic_model.predict(
    X_test_cls
)

logistic_prob = logistic_model.predict_proba(
    X_test_cls
)[:, 1]

print("Logistic Regression")
print("-------------------")

print(
    "Accuracy:",
    accuracy_score(y_test_cls, logistic_pred)
)

print(
    "Precision:",
    precision_score(y_test_cls, logistic_pred)
)

print(
    "Recall:",
    recall_score(y_test_cls, logistic_pred)
)

print(
    "F1:",
    f1_score(y_test_cls, logistic_pred)
)

print(
    "ROC-AUC:",
    roc_auc_score(y_test_cls, logistic_prob)
)
```

```python
rf_classifier = RandomForestClassifier(
    n_estimators=150,
    max_depth=15,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_classifier.fit(
    X_train_cls,
    y_train_cls
)

rf_cls_pred = rf_classifier.predict(
    X_test_cls
)

rf_cls_prob = rf_classifier.predict_proba(
    X_test_cls
)[:, 1]

print("Random Forest Classification")
print("----------------------------")

print(
    "Accuracy:",
    accuracy_score(y_test_cls, rf_cls_pred)
)

print(
    "Precision:",
    precision_score(y_test_cls, rf_cls_pred)
)

print(
    "Recall:",
    recall_score(y_test_cls, rf_cls_pred)
)

print(
    "F1:",
    f1_score(y_test_cls, rf_cls_pred)
)

print(
    "ROC-AUC:",
    roc_auc_score(y_test_cls, rf_cls_prob)
)
```

```python
text_df = df[
    ["Description", "IsReturn"]
].dropna().copy()

text_df["Description"] = (
    text_df["Description"]
    .astype(str)
    .str.lower()
    .str.strip()
)

text_df = text_df[
    text_df["Description"].str.len() > 2
]

print(text_df.shape)
text_df.head()
```

```python
positive = text_df[
    text_df["IsReturn"] == 1
]

negative = text_df[
    text_df["IsReturn"] == 0
]

negative_sample = negative.sample(
    min(len(negative), len(positive) * 2),
    random_state=42
)

nlp_df = pd.concat(
    [positive, negative_sample]
).sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print(nlp_df["IsReturn"].value_counts())
```

```python
tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2),
    stop_words="english"
)

X_text = tfidf.fit_transform(
    nlp_df["Description"]
)

y_text = nlp_df["IsReturn"]

print("TF-IDF shape:", X_text.shape)
```

```python
X_train_text, X_test_text, y_train_text, y_test_text = train_test_split(
    X_text,
    y_text,
    test_size=0.2,
    random_state=42,
    stratify=y_text
)

nlp_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced"
)

nlp_model.fit(
    X_train_text,
    y_train_text
)

nlp_pred = nlp_model.predict(
    X_test_text
)

nlp_prob = nlp_model.predict_proba(
    X_test_text
)[:, 1]
```

```python
print("NLP Model Performance")
print("---------------------")

print(
    "Accuracy:",
    accuracy_score(y_test_text, nlp_pred)
)

print(
    "Precision:",
    precision_score(y_test_text, nlp_pred)
)

print(
    "Recall:",
    recall_score(y_test_text, nlp_pred)
)

print(
    "F1:",
    f1_score(y_test_text, nlp_pred)
)

print(
    "ROC-AUC:",
    roc_auc_score(y_test_text, nlp_prob)
)
```

```python
example_products = [
    "red heart shaped gift box",
    "large ceramic coffee mug",
    "vintage metal decoration"
]

example_vector = tfidf.transform(
    example_products
)

example_prediction = nlp_model.predict(
    example_vector
)

example_probability = nlp_model.predict_proba(
    example_vector
)[:, 1]

for product, prediction, probability in zip(
    example_products,
    example_prediction,
    example_probability
):
    print("Product:", product)
    print("Prediction:", prediction)
    print("Return probability:", round(probability, 3))
    print()
```

```python
def simulate_scenario(
    base_revenue,
    demand_change_percent=0,
    price_change_percent=0
):
    
    demand_multiplier = (
        1 + demand_change_percent / 100
    )
    
    price_multiplier = (
        1 + price_change_percent / 100
    )
    
    simulated_revenue = (
        base_revenue
        * demand_multiplier
        * price_multiplier
    )
    
    revenue_change = (
        simulated_revenue - base_revenue
    )
    
    if revenue_change < 0:
        risk = "High"
    elif revenue_change < base_revenue * 0.05:
        risk = "Medium"
    else:
        risk = "Low"
    
    return {
        "Base Revenue": base_revenue,
        "Simulated Revenue": simulated_revenue,
        "Revenue Change": revenue_change,
        "Risk": risk
    }
```

```python
result = simulate_scenario(
    base_revenue=50000,
    demand_change_percent=20,
    price_change_percent=-5
)

result
```

```python
scenarios = []

scenario_values = [
    ("Baseline", 0, 0),
    ("Demand Growth", 20, 0),
    ("Demand Drop", -20, 0),
    ("Price Increase", 0, 10),
    ("Combined Shock", -20, 10)
]

base_revenue = 50000

for name, demand_change, price_change in scenario_values:
    
    result = simulate_scenario(
        base_revenue,
        demand_change,
        price_change
    )
    
    scenarios.append({
        "Scenario": name,
        "Revenue": result["Simulated Revenue"],
        "Change": result["Revenue Change"],
        "Risk": result["Risk"]
    })

scenario_df = pd.DataFrame(scenarios)

scenario_df
```
